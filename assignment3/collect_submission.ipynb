```python
# PDF export needs system-level Pandoc and XeLaTeX; see README.md.
import os
import shutil
import subprocess

required_tools = ("bash", "zip", "pandoc", "xelatex")
missing_tools = [name for name in required_tools if shutil.which(name) is None]
if missing_tools:
    raise RuntimeError(
        f"Missing system tools: {', '.join(missing_tools)}. "
        "See the PDF export instructions in README.md."
    )
# Ensure the collection script uses this notebook's Python and Jupyter environment.
export_env = os.environ.copy()
export_env["PATH"] = str(Path(sys.executable).parent) + os.pathsep + export_env.get("PATH", "")
(ASSIGNMENT_DIR / "cs231n" / "saved").mkdir(exist_ok=True)
subprocess.run(["bash", "collectSubmission.sh"], cwd=ASSIGNMENT_DIR, env=export_env, check=True)
```
